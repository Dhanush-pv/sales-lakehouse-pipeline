# 03 Data-quality checks: source to Bronze
Fails the workflow if the freshly staged batch is invalid. Bronze is append-only, so duplicate keys there are expected and are **not** checked.

```python
from sales_lakehouse_pipeline import quality

staged = spark.table("staging.sales_incremental")
bronze = spark.table("bronze.sales")

results = [
    quality.check_row_count(bronze, "bronze.sales", min_rows=1),
    quality.check_no_nulls(staged, "staging.sales_incremental", ["customer_id", "updated_at"]),
]
# An empty staging table is valid (nothing changed in the source), so no row-count check on it.
quality.enforce(results)
```
