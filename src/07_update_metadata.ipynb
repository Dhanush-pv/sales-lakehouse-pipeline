# 07 Update watermark
Runs last, so a failed run never advances the watermark.

```python
max_ts = spark.table("staging.sales_incremental").agg({"updated_at": "max"}).collect()[0][0]

if max_ts is None:
    print("No new data found. Watermark not updated.")
else:
    spark.sql(f"""
    UPDATE metadata.pipeline_metadata
    SET watermark = TIMESTAMP'{max_ts}'
    WHERE pipeline_name = 'sales_pipeline'
    """)
    print(f"Watermark updated to {max_ts}")
```
