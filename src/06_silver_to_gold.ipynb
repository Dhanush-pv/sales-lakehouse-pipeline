# 06 Silver to Gold
Builds the business aggregate: total sales per region.

```python
%sql
CREATE OR REPLACE TABLE gold.sales_summary AS
SELECT region, SUM(amount) AS total_sales
FROM silver.sales
GROUP BY region
```

```python
from sales_lakehouse_pipeline import quality

gold = spark.table("gold.sales_summary")
quality.enforce([
    quality.check_row_count(gold, "gold.sales_summary", min_rows=1),
    quality.check_unique_key(gold, "gold.sales_summary", ["region"]),
])
```
